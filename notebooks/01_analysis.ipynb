```python
import pandas as pd
```

```python
news=pd.read_csv("news.csv",encoding='latin1')
```

```python
print(news.columns)
```

```python
news = news.drop(columns=['Description', 'Author', 'Content', 'Keywords',
    'Unnamed: 6', 'Unnamed: 7', 'Unnamed: 8', 'Unnamed: 9',
    'Unnamed: 179', 'Unnamed: 180', 'Unnamed: 181', 'Unnamed: 182',
    'Unnamed: 183', 'Unnamed: 184', 'Unnamed: 185', 'Unnamed: 186',
    'Unnamed: 187', 'Unnamed: 188'])
```

```python
news=news.iloc[:,:2]
```

```python
print(news.columns)
print(news.shape)
```

```python
print(news.isnull().sum())
```

```python
news = news.dropna()
news = news.drop_duplicates()
news['Title'] = news['Title'].str.strip()
```

```python
print(news.info())
print(news.describe())
print(news.head())
```

```python
grouped_news = (news.groupby("Date")["Title"].apply(list).reset_index())
```

```python
print(grouped_news.head())
```

```python
grouped_news = grouped_news.iloc[1:]
```

```python
print(grouped_news.head())
print(grouped_news.shape)
```

```python
grouped_news['Date'] = pd.to_datetime(
    grouped_news['Date'],
    format='%d-%m-%Y',
    errors='coerce'
)
```

```python
print(grouped_news['Date'].isnull().sum())
print(grouped_news.head())
```

```python
stock = pd.read_csv("nifty50.csv", encoding='cp1252')
```

```python
print(stock.columns)
print(stock.head())
```

```python
stock.columns = [
    'date',
    'open',
    'high',
    'low',
    'close',
    'shares_traded',
    'turnover_cr'
]
```

```python
stock['shares_traded'] = stock['shares_traded'].astype(float)
stock['date'] = pd.to_datetime(stock['date'], dayfirst=True, errors='coerce')
```

```python
print(stock.dtypes)
print(stock['date'].isnull().sum())
print(stock.head())
```

```python
merged = pd.merge(stock, grouped_news, left_on='date', right_on='Date', how='inner')
```

```python
print(merged.isnull().sum())
print(merged.shape)
print(merged.head())
```

```python
print(merged.columns)
print(merged.head())
```

```python
print(merged.columns)
print(merged.head())
```

```python
merged['daily_return'] = ((merged['close'] - merged['open']) / merged['open']) * 100

merged['movement'] = merged['daily_return'].apply(lambda x: 1 if x > 0 else 0)

merged['volatility_pct'] = ((merged['high'] - merged['low']) / merged['open']) * 100
```

```python
print(merged[['date', 'open', 'close', 'daily_return', 'movement', 'volatility_pct']].head())
print(merged[['daily_return', 'volatility_pct']].describe())
print(merged['movement'].value_counts())
```

```python
print(merged.columns)
print(merged[['date', 'open', 'close', 'daily_return', 'movement', 'volatility_pct']].head())
print(merged['movement'].value_counts())
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
```

```python
sample_text = merged['Title'].iloc[0]

print(sample_text)
print(analyzer.polarity_scores(sample_text))
```

```python
merged['sentiment_score'] = merged['Title'].apply(
    lambda text: analyzer.polarity_scores(str(text))['compound']
)
```

```python
def get_sentiment_label(score):
    if score >= 0.05:
        return 'positive'
    elif score <= -0.05:
        return 'negative'
    else:
        return 'neutral'

merged['sentiment_label'] = merged['sentiment_score'].apply(get_sentiment_label)
```

```python
print(merged[['date', 'Title', 'sentiment_score', 'sentiment_label']].head())
print(merged['sentiment_label'].value_counts())
print(merged['sentiment_score'].describe())
```

```python
print(type(merged['Title'].iloc[0]))
print(merged['Title'].iloc[0])
```

```python
merged['news_count'] = merged['Title'].apply(
    lambda x: len(x) if isinstance(x, list) else 1
)
```

```python
print(merged[['date', 'Title', 'news_count']].head())
print(merged['news_count'].describe())
```

```python
merged = merged.sort_values('date').reset_index(drop=True)
```

```python
merged['prev_day_sentiment'] = merged['sentiment_score'].shift(1)
```

```python
merged['prev_day_sentiment'] = merged['prev_day_sentiment'].fillna(0)
```

```python
print(merged[['date', 'sentiment_score', 'prev_day_sentiment', 'news_count']].head(10))
print(merged[['sentiment_score', 'prev_day_sentiment', 'news_count']].describe())
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=merged, x='sentiment_score', y='daily_return', hue='movement')
plt.title('Sentiment Score vs Daily Return')
plt.xlabel('Sentiment Score')
plt.ylabel('Daily Return (%)')
plt.show()
```
